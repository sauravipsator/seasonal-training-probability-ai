# 03 - Exploratory data analysis
Use this notebook to inspect station demand, seasonal patterns, vendor coverage and train-route concentration before retraining.

```python
import sys
sys.path.append('..')
from train_station_demand_forecast import order_rows, daily_orders

orders = order_rows()
daily = daily_orders(orders)
daily.groupby('station_name').order_count.sum().nlargest(10)
```

```python
monthly = daily.assign(month=daily.date.dt.to_period('M')).groupby('month').order_count.sum()
ax = monthly.plot(figsize=(12, 4), title='Delivered orders by month', ylabel='Orders')
ax.figure.tight_layout()
```

```python
route_concentration = orders.groupby(['station_code', 'train_no', 'train_name']).size().rename('orders').reset_index()
route_concentration.sort_values('orders', ascending=False).head(15)
```
